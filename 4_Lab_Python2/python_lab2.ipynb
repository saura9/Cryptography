6. Read cryptology_no.txt and count the number of:

Uppercase letters
Lowercase letters
Digits
Spaces
Special characters

```python
file = open("cryptology_no.txt", "r")
content = file.read()
uppercase = 0
lowercase = 0
digits = 0
spaces = 0
special = 0

for ch in content:
    if ch.isupper():
        uppercase += 1
    elif ch.islower():
        lowercase += 1
    elif ch.isdigit():
        digits += 1
    elif ch == " ":
        spaces += 1
    elif ch != "\n":
        special += 1

print("Uppercase letters :", uppercase)
print("Lowercase letters :", lowercase)
print("Digits            :", digits)
print("Spaces            :", spaces)
print("Special characters:", special)

file.close()
```

6. Write a Python program to identify and extract the port number from statements of the form Port 443 connection established.


```python
import re

file = open("security_data.txt", "r")
content = file.read()

pattern = r'\bPort\s+(\d+)\s+connection established\b'

match = re.search(pattern, content)

if match:
    print("Port number:", match.group(1))
else:
    print("Port number not found.")

file.close()
```

7.Write a Python program to read the contents of security_data.txt and use regular expressions to extract IP-address-like patterns, email addresses, session IDs, and hexadecimal sequences. Store the identified patterns in a new file named pattern_results.txt.


```python
import re

# Read security_data.txt
file = open("security_data.txt", "r")
content = file.read()
file.close()

# Regular expressions
ip_pattern = r'\b(?:\d{1,3}\.){3}\d{1,3}\b'
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
session_pattern = r'\bSEC-\d{4}-\d{4}\b'
hex_pattern = r'\b[0-9a-fA-F]{8,}\b'

# Extract patterns
ips = re.findall(ip_pattern, content)
emails = re.findall(email_pattern, content)
sessions = re.findall(session_pattern, content)
hex_values = re.findall(hex_pattern, content)

# Store results
file = open("pattern_results.txt", "w")

file.write("IPv4 Addresses:\n")
for ip in ips:
    file.write(ip + "\n")

file.write("\nEmail Addresses:\n")
for email in emails:
    file.write(email + "\n")

file.write("\nSession IDs:\n")
for session in sessions:
    file.write(session + "\n")

file.write("\nHexadecimal Sequences:\n")
for value in hex_values:
    file.write(value + "\n")

file.close()

print("Patterns extracted and stored in pattern_results.txt")
```
