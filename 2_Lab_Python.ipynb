```python
n = int(input("Enter a number: "))
if n < 2:
    print("Not prime number")
else:
    prime = True
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            prime = False
            break

    if prime:
        print("Prime Number")
    else:
        print("Not a prime number")
```

```python
num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))
while num2 != 0:
    num1, num2 = num2, num1 % num2

print("GCD:", num1)
if num1 == 1:
    print("Numbers are coprime")
else:
    print("Numbers are not coprime")
```

```python
b = int(input("integer b: "))
inverse = -1
for x in range(1, 26):
    if (b * x) % 26 == 1:
        inverse = x
        break

if inverse != -1:
    print("Multiplicative inverse of", b, "= ", inverse)
else:
    print("Multiplicative inverse does not exist.")
```
