1. Caseser Cipher Encryption and Decryption

```python
pt= input("Enter plain text: ")

import string
key=3
cipher=[]
for ch in pt:
    if(ch in string.ascii_uppercase):
        ct = (ord(ch) - ord('A') + key) % 26 + ord('A')
        cipher.append(chr(ct))
        
    elif(ch in string.ascii_lowercase):
        ct = (ord(ch) - ord('a') +key) %26 + ord('a')
        cipher.append(chr(ct))
    else:
        cipher.append(ch)

cipher_txt= "".join(cipher)
print(cipher_txt)
    
```

2. Shift Cipher Encryption and Decryption

```python
def shift_cipher(text, key, mode="encrypt"):
    
    if mode == "decrypt":
        key = -key
    
    result = []
    
    for ch in text:
        if ch.isupper():
            p = (ord(ch) - ord('A') + key) % 26 + ord('A')
            result.append(chr(p))
        elif ch.islower():
            p = (ord(ch) - ord('a') + key) % 26 + ord('a')
            result.append(chr(p))
        else:
            result.append(ch)
            
    return "".join(result)

secret_key = int(input("Enter secret key bw 0-25 : "))
secret_key%=26
message= input("Enter the message : ")

cipher_text = shift_cipher(message, secret_key, "encrypt")
print(f"Encrypted: {cipher_text}") 

decrypted_msg = shift_cipher(cipher_text, secret_key, "decrypt")
print(f"Decrypted: {decrypted_msg}")
```

5. ATBASH Encryption and Decryption

```python
def atbash(text):
    result = []
    for char in text:
        if char.isupper():
            result.append(chr(ord('Z') - (ord(char) - ord('A'))))
        elif char.islower():
            result.append(chr(ord('z') - (ord(char) - ord('a'))))
        else:
            result.append(char)
    return "".join(result)

message = input("Enter Plain text to be encrypted using Atbash cipher: ")
encrypted = atbash(message)
decrypted = atbash(encrypted)

print("Atbash Encrypted:", encrypted)
print("Atbash Decrypted:", decrypted)
```

6. AFFINE CIPHER


```python
def _mod_inverse(a, m):
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    raise ValueError(f"Modular inverse does not exist for a={a} and m={m}")

def affine_encrypt(text, a, b):
    result = []
    for char in text:
        if char.isalpha():
            ascii_conv = ord('A') if char.isupper() else ord('a')
            x = ord(char) - ascii_conv
            encrypted = (a * x + b) % 26
            result.append(chr(encrypted + ascii_conv))
        else:
            result.append(char)
    return "".join(result)

def affine_decrypt(text, a, b):
    result = []
    a_inv = _mod_inverse(a, 26)
    for char in text:
        if char.isalpha():
            ascii_conv = ord('A') if char.isupper() else ord('a')
            y = ord(char) - ascii_conv
            decrypted = (a_inv * (y - b)) % 26
            result.append(chr(decrypted + ascii_conv))
        else:
            result.append(char)
    return "".join(result)

message = input("Enter text to encrypt - affine cipher ")
a_key, b_key = 7, 9  # 'a' must be coprime to 26
encrypted = affine_encrypt(message, a_key, b_key)
decrypted = affine_decrypt(encrypted, a_key, b_key)

print("Affine Encrypted:", encrypted)
print("Affine Decrypted:", decrypted)
```
