1. Write a python code to perform One Time Pad


```python
def otp_encrypt(text, key):
    result = ""

    for i in range(len(text)):
        x = ord(text[i]) - ord('A')
        k = ord(key[i]) - ord('A')

        result += chr((x + k) % 26 + ord('A'))

    return result


def otp_decrypt(cipher, key):
    result = ""

    for i in range(len(cipher)):
        c = ord(cipher[i]) - ord('A')
        k = ord(key[i]) - ord('A')

        result += chr((c - k) % 26 + ord('A'))

    return result


text = input("Enter plaintext: ").upper()
key = input("Enter key: ").upper()

if len(text) != len(key):
    print("Error: Key must be same length as plaintext.")
else:
    cipher = otp_encrypt(text, key)

    print("Ciphertext:", cipher)
    print("Decrypted text:", otp_decrypt(cipher, key))
```

2. Write a Python code to break the shift cipher "MCI RWR WH" (Perform a brute-force attack by displaying all 26 possible plaintexts.)


```python
cipher = "MCI RWR WH"

for key in range(26):
    plaintext = ""

    for ch in cipher:
        if ch.isalpha():
            x = ord(ch) - ord('A')
            plaintext += chr((x - key) % 26 + ord('A'))
        else:
            plaintext += ch

    print("Key", key, ":", plaintext)
```

3. Write a Python code to encrypt Rail Fence

```python
def rail_fence_encrypt(text, rails):
    if rails == 1:
        return text

    fence = [[] for _ in range(rails)]

    row = 0
    direction = 1

    for ch in text:
        fence[row].append(ch)

        if row == 0:
            direction = 1
        elif row == rails - 1:
            direction = -1

        row += direction

    result = ""

    for rail in fence:
        result += ''.join(rail)

    return result


text = input("Enter plaintext: ")
rails = int(input("Enter number of rails: "))

cipher = rail_fence_encrypt(text, rails)
print("Plain text:", text)
print("Depth:", rails)

print("Encrypted text:", cipher)
```
