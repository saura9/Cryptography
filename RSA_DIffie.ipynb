1. write a python script to implement RSA algorithm 



```python

# RSA Algorithm - Text Encryption with User Input

import math

p = int(input("Enter first prime number (p): "))
q = int(input("Enter second prime number (q): "))

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, math.isqrt(n) + 1):
        if n % i == 0:
            return False
    return True

if not is_prime(p) or not is_prime(q) or p == q:
    print("Enter two distinct prime numbers.")
    raise SystemExit

n = p * q
phi = (p - 1) * (q - 1)

# Find public key exponent
e = 2
while e < phi:
    if math.gcd(e, phi) == 1:
        break
    e += 1

# Find private key exponent
d = pow(e, -1, phi)

print("\nPublic Key:", (e, n))
print("Private Key:", (d, n))

message = input("\nEnter message to encrypt: ")

# Encrypt each character separately
encrypted = [pow(ord(ch), e, n) for ch in message]
print("Encrypted message:", encrypted)

# Decrypt each character
decrypted = ''.join(chr(pow(c, d, n)) for c in encrypted)
print("Decrypted message:", decrypted)
```

```python

# Diffie-Hellman Key Exchange with User Input

# Public parameters
p = int(input("Enter a prime number (p): "))
g = int(input("Enter a generator (g): "))

if p < 3:
    print("Error: Prime number must be at least 3.")
    raise SystemExit


def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

if not is_prime(p):
    print("Error: p must be prime.")
    raise SystemExit

if not 1 < g < p:
    print("Error: g must be between 2 and p - 1.")
    raise SystemExit

# Private keys chosen by Alice and Bob
a = int(input("Enter Alice's private key: "))
b = int(input("Enter Bob's private key: "))

if not (1 <= a <= p - 2 and 1 <= b <= p - 2):
    print("Error: Private keys must be between 1 and p - 2.")
    raise SystemExit

# Public keys
A = pow(g, a, p)
B = pow(g, b, p)

print("\nAlice's public key:", A)
print("Bob's public key:", B)

# Calculate shared secret keys
key_alice = pow(B, a, p)
key_bob = pow(A, b, p)

print("Alice's shared key:", key_alice)
print("Bob's shared key:", key_bob)

# Verify the shared keys
if key_alice == key_bob:
    print("Key exchange successful!")
else:
    print("Key exchange failed!")
```
